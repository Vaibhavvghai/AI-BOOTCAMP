```python
import numpy as np

# 1. Create a 2D array (Matrix)
matrix = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

print("Original 2D Array:")
print(matrix)

# 2. Find the average of each row (axis=1 collapses columns)
row_averages = np.mean(matrix, axis=1)
print("\nAverage of each row:")
print(row_averages)

# 3. Find the average of each column (axis=0 collapses rows)
column_averages = np.mean(matrix, axis=0)
print("\nAverage of each column:")
print(column_averages)

# 4. Calculate the standard deviation of the entire array
std_deviation = np.std(matrix)
print(f"\nOverall Standard Deviation: {std_deviation:.2f}")
```

```python
import numpy as np

# 1. Create a 2D array (Matrix)
random_matrix = np.random.randint(10,100, size = (10,10))

print("Original 2D Array:")
print(random_matrix)

# 2. Find the average of each row (axis=1 collapses columns)
row_averages = np.mean(random_matrix, axis=1)
print("\nAverage of each row:")
print(row_averages)

# 3. Find the average of each column (axis=0 collapses rows)
column_averages = np.mean(random_matrix, axis=0)
print("\nAverage of each column:")
print(column_averages)

# 4. Calculate the standard deviation of the entire array
std_deviation = np.std(random_matrix)
print(f"\nOverall Standard Deviation: {std_deviation:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Amplitude
A = 5

# Generate x values from 0 to 2π
x = np.linspace(0, 100, 1000)

# Generate sine wave
y = A * np.sin(x)

# Plot the sine wave
plt.plot(x, y)

# Labels and title
plt.xlabel("x")
plt.ylabel("Amplitude")
plt.title("Sine Wave")

# Show grid
plt.grid()

# Display graph
plt.show()
```

```python
import matplotlib.pyplot as plt

# Height groups
height = ['140-149', '150-159', '160-169', '170-179', '180-189', '190-199']

# Number of people
people = [5, 12, 20, 25, 15, 5]

# Create horizontal bar graph
plt.barh(height, people)

plt.xlabel("Number of People")
plt.ylabel("Height (cm)")
plt.title("Height-wise People Graph")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate values from 0 to 2π
t = np.linspace(0, 100, 100)

# X-axis = sine wave
x = np.sin(t)

# Y-axis = cosine wave
y = np.cos(t)

# Plot graph
plt.plot(x, y)

plt.xlabel("Sine Wave")
plt.ylabel("Cosine Wave")
plt.title("Sine vs Cosine")

plt.grid()
plt.show()
```
